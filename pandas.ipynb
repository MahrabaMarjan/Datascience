```python
import pandas as pd
sales= pd.Series(
    [1200,850,1500,950,2000],
    index=["Monday","Tuesday","Wednesday","Thursday","Friday"]
)
print(sales)
print("maximum sale=",sales.max())

print("minimum sale=",sales.min())
print(sales[sales<1000])
```

```python
print(sales.sum())
```

```python
import pandas as pd #unique is used to access unique elements
data=[10,20,20,20,30,40,50,50,40,60]
r=pd.Series(data)
print(r)
print(r.unique())
print(r.nunique()) # count of unique no.
```

```python
print(r.value_counts())
```

```python
data=[10,20,None,20,30,40,50,None,60]
t=pd.Series(data)
print(t.isna())  #isna will give false for where the value is not null and true for a null value
print(t.notna()) #notna will give true for where the value is not null and false for a null value
```

```python
data=[10,20,None,20,30,40,50,None,60]
t=pd.Series(data)
print(t.isnull()) #isnull will give false for where the value is not null and true for a null value
print(t.notnull()) #notnull will give true for where the value is not null and false for a null value
```
