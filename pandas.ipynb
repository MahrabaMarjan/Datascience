```python
import pandas as pd
marks=pd.Series(
    [78,45,89,67,92,45,56,78,None,34],
    index=["Anu","Rahul","Meera","Arun","Neha",
           "Vishnu","Diya","Akhil","Riya","Kiran"]
)
```

```python
#Find the number of students and total marks
print(len(marks.index))
print(marks.sum())
```

```python
#Find the average, highest, and lowest marks
print("average mark=",marks.mean())
print("highest mark=",marks.max())
print("lowest  mark=",marks.min())
```

```python
#Find the name of the student with the highest mark
print("student with the highest mark =",marks.idxmax())
```

```python
#Check and count the missing values
print(marks.isna())
print(marks.isna().sum())
```

```python
#Replace the missing value with 0
print(marks.fillna(0))
```

```python
#Find the unique marks and the number of unique marks.
print(marks.unique())
print(marks.nunique())
print(marks.value_counts())
```

```python
#Find students who scored above 70
print(marks[marks>70])
```

```python
#Sort the marks in descending order.
print(marks.sort_values(ascending=False))
```
