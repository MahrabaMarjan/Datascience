# MATPLOTLIB
it is a python library used to create graphs and visualizing data

---
**PYPLOT**
 * pyplot is a module in matplotlib used to create and customize new features to graphs in python

```python
import matplotlib.pyplot as plt
x= [1,2,3,4,5]
y=[2,4,6,8,10]

plt.plot(x,y)
plt.show()
```

```python
Month = ["Jan","Feb","mar","apr","may"]
Sales = [20,30,25,40,50]
plt.plot(Month,Sales)
plt.show()
```

```python
Month = ["Jan","Feb","mar","apr","may"]
Sales = [1000,2000,2500,1500,3500]
Profit = [5,15,50,10,90]

plt.plot(Month,Sales,label="Sales") # label indicates the different name of labelled graphs
plt.plot(Month,Profit,label="Profit")
plt.legend() # to display the legends
plt.show()
```

```python
import matplotlib.pyplot as plt

Month = ["Jan","Feb","mar","apr","may"]
Sales = [20,30,25,40,50]
plt.plot(Month,Sales)


plt.xlabel("MONTHS")
plt.ylabel("SALES")
plt.title("Monthly sales")
plt.grid()
plt.plot(
    Month,
    Sales,
    marker="*",
    color="blue",
    linestyle="--"
)
plt.show()
```

```python
import matplotlib.pyplot as plt

Month = ["Jan","Feb","mar","apr","may"]
Sales = [20,30,25,40,50]
plt.bar(Month,Sales)


plt.xlabel("MONTHS")
plt.ylabel("SALES")
plt.title("Monthly sales")
plt.grid(False)
plt.bar(
    Month,
    Sales,
    color="#FF0040"
)
plt.show()
```

```python
Month = ["Jan","Feb","mar","apr","may"]
Sales = [20,30,25,40,50]
plt.scatter(Month,Sales)


plt.xlabel("MONTHS")
plt.ylabel("SALES")
plt.title("Monthly sales")
plt.grid(True)
plt.scatter(
    Month,
    Sales,
    color="#000000"
)
plt.show()
```

```python
Marks=[45,30,20,30,50,60.78,99,77]
plt.hist(Marks)
plt.title=("Report Card")

plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.grid(True)
plt.show()
```

```python
categories=["Food","Travel","Shopping","Others"]
amount=[5000,2000,3000,1500]

plt.pie(amount,labels=categories)

plt.legend()


plt.title("Monthly Expense")
plt.show()
```

```python
fig, ax = plt.subplots(2,1)
ax[0].plot([1,2,3],[10,20,30])
ax[1].bar(["travel","food","shopping"],[400,600,800])
plt.show()
```
